# **Sorting**
- Sorting along an axis axis=1 ----> row and axis=0----->col

```python
a=np.array([[5,4,6],[2,8,2],[8,9,20]])
print(a)
b=np.sort(a,axis=0)
```

```python
print(b) # ------> sorted colwise
```

```python
np.sort(a,axis=1)
```

```python
a
```

# -----> when axis is not specified the it is row wise sorting

```python
np.sort(a) # rowwise sorting
```
