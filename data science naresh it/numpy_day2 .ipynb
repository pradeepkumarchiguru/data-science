# **random.randint**
-  Return random integer in range [a, b], including start but  end points is exluded.

- **Syntax**: np.random.randint(start,end, no_of_samples)

```python
# importing random module
import random
# i want 5 random number
a=[]
for i in range(5):
    a.append(random.randint(1,6))
print(a)
```

```python
np.random.randint( 1,5,10)
```

```python
def random_randint(start_val, end_val, num_samples):
  import random
  l=[]
  for _ in range(num_samples):
    y=random.randint(start_val, end_val)
    l.append(y)
  a=np.array(l)
  return a
```

# **Backend code**
- this is the backendcode of np.random.randint

```python
def random_randint(star,end,number_of_samples):
    l=[]
    for i in range(number_of_samples):
       l.append(random.randint(star,end))
    a=np.array(l)
    return a


```

```python
random_randint(1,5,10)
```

# **Accesing a multiple values based on condition [mask indexing  or fancy indexing ]**
- **Syntax** array[condition]
- first  it will check condition , the items which satisfies the condition , tat item will be select  

```python
s=np.array([45,46,47,48])
s
```

- to get the boolean condition

```python
s>45
```

**backend code**

```python
l=[]
for i in s:
  l.append( i>45)
np.array(l)
```

-  to  get data based on  condition
- array[condition]

```python
s
```

```python
s[s>47]
```

**backend code**

```python
l=[]
for i in s:
  if i>47:
    l.append(i)
np.array(l)
```
